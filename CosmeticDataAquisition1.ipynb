Dataset Description

This dataset contains 1,472 cosmetic and skincare products collected from Kaggle. The dataset includes information on product category, brand, product name, price, product rating, ingredients and suitability for different skin types. Initial exploration found no missing values in the dataset. This dataset will be combined with additional information collected from the Ulta Beauty website.
