Web Scraping Ulta Beauty

To complement the Kaggle Cosmetics Dataset, additional Infornation will be collected from the Ulta Beauty Website. The data extracted will include..... This information will allow the Investigation of whether more popular products receive higher ratings

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.ulta.com/shop/skin-care/all"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title)
```
